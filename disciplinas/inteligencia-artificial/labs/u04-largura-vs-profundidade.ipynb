## 1. O grafo

Usaremos um grafo não direcionado representado como dicionário de adjacências. A ordem dos vizinhos em cada lista importa: é ela que define a ordem em que os filhos são gerados.

```python
grafo = {
    "A": ["B", "C"],
    "B": ["A", "D", "E"],
    "C": ["A", "F", "G"],
    "D": ["B", "H"],
    "E": ["B", "H", "I"],
    "F": ["C", "J"],
    "G": ["C", "J"],
    "H": ["D", "E", "K"],
    "I": ["E", "K"],
    "J": ["F", "G", "L"],
    "K": ["H", "I", "L"],
    "L": ["J", "K"],
}
print(f"{len(grafo)} nós no grafo")
```

## 2. A busca genérica

A função abaixo mantém uma **fronteira** (nós gerados e ainda não examinados) e um conjunto de **visitados**. Cada entrada da fronteira guarda o caminho completo até o nó.

**Sua tarefa:** complete a linha marcada com `TODO`. Se `modo` for `"largura"`, remova o **primeiro** elemento da fronteira (fila). Se for `"profundidade"`, remova o **último** (pilha).

```python
from collections import deque

def busca(grafo, inicio, objetivo, modo="largura"):
    fronteira = deque([[inicio]])
    visitados = set()
    examinados = 0

    while fronteira:
        # TODO: remova o próximo caminho da fronteira de acordo com o modo
        # Dica: deque tem os métodos popleft() e pop()
        caminho = None
        if caminho is None:
            raise NotImplementedError("Complete a linha do TODO")

        no = caminho[-1]
        if no in visitados:
            continue
        visitados.add(no)
        examinados += 1

        if no == objetivo:
            return caminho, examinados

        for vizinho in grafo[no]:
            if vizinho not in visitados:
                fronteira.append(caminho + [vizinho])

    return None, examinados
```

## 3. Comparando as estratégias

Execute a célula abaixo depois de completar o `TODO`.

```python
for objetivo in ["L", "I"]:
    for modo in ["largura", "profundidade"]:
        caminho, examinados = busca(grafo, "A", objetivo, modo)
        print(f"objetivo {objetivo} | {modo:<12} | examinados: {examinados:>2} | caminho: {' -> '.join(caminho)}")
    print()
```
